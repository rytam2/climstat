```python
path = '/data/keeling/a/rytam2/climstat/local/data/'
hi_tot = pd.read_csv(path+'ERA5_heat_index_totalstats_table_2016_2025_ILCounties.csv') 
hi_yearly = pd.read_csv(path+'ERA5_heat_index_yearlystats_table_2016_2025_ILCounties.csv')
hi_daily = pd.read_csv(path+'ERA5_heat_index_daily_2016_2025_ILCounties_06172025.csv')
hi_mean = pd.read_csv(path+'ERA5_heat_index_2016_2025_mean_ILCounties.csv')# remove path+
```

```python
hi_tot = hi_tot.rename(columns={'HI Days above 90': 'Total HI Days above 90 (Day)',
                    'HI Hours above 90': 'Total HI Hours above 90 (Hr)',
                      'HI Days above 104': 'Total HI Days above 104 (Day)',
                    'HI Hours above 104': 'Total HI Hours above 104 (Hour)',
                    'HI JJA Daily Av. High':'2016-2025 HI JJA Daily Av. High (F)',
                    'HI JJA Daily Av. Low': '2016-2025 HI JJA Daily Av. Low (F)'},
                    ).drop('Unnamed: 0', axis=1).set_index('County')

# hi_summary.append()
hi_avg = hi_yearly.drop(['Unnamed: 0','HI JJA Av. Daily High','HI JJA Av. Daily Low','Year','FIPS'], axis=1).groupby('County').mean(['HI Days above 90','HI Hours above 90',
                                                            'HI Days above 104','HI Hours above 104'])\
                                                .rename(columns={'HI Days above 90': '2016-2025 HI Avg. Days above 90 (Day)',
                                                                'HI Hours above 90': '2016-2025 HI Avg. Hours above 90 (Hr)',
                                                                'HI Days above 104': '2016-2025 HI Avg. Days above 104 (Day)',
                                                                'HI Hours above 104': '2016-2025 HI Avg. Hours above 104 (Hour)'})
# Record Max
hi_daily['Day'] = pd.to_datetime(hi_daily['Day'])
hi_record_max = hi_daily[hi_daily['Day'].dt.month.isin([6, 7, 8])].groupby('County').max()['HI max'].rename('HI 2016-2025 Record Max')
hi_record_time = hi_daily[hi_daily['Day'].dt.month.isin([6, 7, 8])].groupby('County').max()['Day'].rename('HI 2016-2025 Record Date')

left = hi_tot.iloc[:, :5]
right = hi_tot.iloc[:, 5:]
hi_summary = pd.concat([left,hi_avg,right,hi_record_max,hi_record_time], axis=1)
hi_summary.keys()

# hi_summary.to_csv('summarystats_hi_by_county_2016-2025.csv')
```
